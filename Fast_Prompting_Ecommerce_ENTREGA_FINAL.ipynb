### Resumen
Este proyecto aborda una problemática frecuente en pequeños e-commerce: la necesidad de producir contenido promocional de forma constante sin disponer siempre de tiempo, conocimientos de marketing o presupuesto para tercerizar la tarea.

La propuesta consiste en desarrollar una prueba de concepto (POC) que utilice técnicas de Fast Prompting para generar dos tipos de contenido complementarios a partir de la información de un producto: una publicación promocional mediante un modelo **texto → texto** y un prompt optimizado para producir una pieza visual mediante un modelo **texto → imagen**. La implementación se desarrolla en una Jupyter Notebook reutilizable y orientada a reducir ambigüedad, mejorar consistencia y evitar consultas innecesarias.

# 1. Introducción

## 1.1 Nombre del proyecto
**Fast Prompting aplicado a la automatización de contenido para e-commerce.**

## 1.2 Presentación del problema a abordar
Los pequeños e-commerce necesitan mantener una presencia activa en redes sociales y tiendas online. Sin embargo, generar publicaciones atractivas de manera frecuente requiere tiempo, criterios de comunicación, conocimientos de marketing y capacidad para producir piezas visuales.

## 1.3 Desarrollo de la propuesta de solución
La solución propuesta utiliza Inteligencia Artificial generativa para transformar una ficha estructurada de producto en dos recursos de marketing:

1. **Texto → texto:** generación de una publicación promocional adaptada al canal, público, tono y objetivo.
2. **Texto → imagen:** generación de una pieza visual coherente con el mismo producto y público objetivo.

Para mejorar la calidad de las respuestas se aplican técnicas de Fast Prompting basadas en la organización explícita de **rol, contexto, tarea, restricciones y formato de salida**.

## 1.4 Justificación de la viabilidad
El proyecto es técnicamente viable porque utiliza una Jupyter Notebook, Python y una integración simple con la API de OpenAI para texto. La configuración de la API se realiza mediante una variable de entorno, evitando incluir credenciales en el repositorio.

Para la generación de imagen no es obligatorio utilizar una API paga. La consigna permite utilizar herramientas gratuitas externas como NightCafe u otras alternativas. En ese caso, el prompt se copia manualmente en la herramienta elegida y la imagen obtenida se incorpora posteriormente al repositorio.

La POC trabaja con un único producto de ejemplo y prompts reutilizables, por lo que puede completarse con recursos limitados y sin necesidad de desarrollar una aplicación completa.

## Técnicas de Fast Prompting utilizadas
La estrategia principal organiza cada prompt en cinco componentes:

- **Rol:** define el perfil que debe adoptar el modelo.
- **Contexto:** aporta la información necesaria del producto.
- **Tarea:** especifica qué debe producirse.
- **Restricciones:** limita respuestas no deseadas y evita invenciones.
- **Formato:** define cómo debe presentarse la salida.

Estas técnicas reducen ambigüedad, aumentan consistencia y facilitan la reutilización del prompt.

# 5. Configuración
La API key se obtiene desde una variable de entorno. Esto evita almacenar credenciales dentro del código o del repositorio.

La notebook puede recorrerse con `usar_api = False` sin realizar consultas.

```python
import os
from openai import OpenAI

api_key = os.getenv("OPENAI_API_KEY")
usar_api = False

cliente = OpenAI(api_key=api_key) if api_key else None

print("API configurada:", api_key is not None)
print("Ejecutar API:", usar_api)
```

# 6. Datos de entrada

La información del producto se mantiene separada de los prompts para que el sistema sea reutilizable.

```python
producto = {
    "nombre": "Zapatillas Urban Street",
    "caracteristicas": "zapatillas urbanas cómodas, diseño moderno y suela liviana",
    "publico": "jóvenes de 18 a 30 años interesados en moda urbana",
    "canal": "Instagram",
    "tono": "cercano, moderno y dinámico",
    "objetivo": "generar interés y dirigir potenciales clientes a la tienda online"
}

for clave, valor in producto.items():
    print(f"{clave.capitalize()}: {valor}")
```

# 7. Implementación texto → texto

## 7.1 Prompt básico
El primer prompt funciona como línea de base. Incluye el objetivo principal pero deja varias decisiones abiertas al modelo.

```python
def crear_prompt_basico(datos):
    return (
        f"Crea una publicación para {datos['canal']} promocionando "
        f"{datos['nombre']}. Características: {datos['caracteristicas']}."
    )

prompt_basico = crear_prompt_basico(producto)
print(prompt_basico)
```

## 7.2 Fast Prompt estructurado

El segundo prompt incorpora rol, contexto, tarea, restricciones y formato. Esta estructura busca aumentar el control sobre la respuesta.

```python
def crear_fast_prompt(datos):
    return f"""
ROL
Actúa como especialista en marketing digital para pequeños e-commerce.

CONTEXTO
Producto: {datos['nombre']}
Características: {datos['caracteristicas']}
Público objetivo: {datos['publico']}
Canal: {datos['canal']}
Tono: {datos['tono']}
Objetivo: {datos['objetivo']}

TAREA
Redacta una publicación promocional lista para utilizar en {datos['canal']}.

RESTRICCIONES
- Escribe entre 70 y 110 palabras.
- Mantén el tono indicado.
- Destaca únicamente características proporcionadas en el contexto.
- No inventes descuentos, precios, stock, envíos ni beneficios no informados.
- Incluye un llamado a la acción breve.
- Incluye exactamente 5 hashtags relevantes.
- Evita explicaciones sobre cómo construiste la respuesta.

FORMATO DE SALIDA
PUBLICACIÓN:
<texto>

CTA:
<una frase>

HASHTAGS:
<5 hashtags>
""".strip()

prompt_fast = crear_fast_prompt(producto)
print(prompt_fast)
```

## 7.3 Consulta al modelo

Todas las consultas pasan por una misma función. El uso de la API queda desactivado por defecto para evitar costos involuntarios.

```python
def consultar_modelo(prompt):
    if not usar_api:
        return {
            "texto": "[SIMULACIÓN] La API está desactivada.",
            "input_tokens": None,
            "output_tokens": None,
            "total_tokens": None
        }

    if not api_key:
        raise ValueError("No se encontró OPENAI_API_KEY en las variables de entorno.")

    response = cliente.responses.create(
        model="gpt-4o",
        input=prompt
    )

    usage = getattr(response, "usage", None)

    return {
        "texto": response.output_text,
        "input_tokens": getattr(usage, "input_tokens", None) if usage else None,
        "output_tokens": getattr(usage, "output_tokens", None) if usage else None,
        "total_tokens": getattr(usage, "total_tokens", None) if usage else None
    }
```

## 7.4 Experimento: comparación de prompts

Se utilizan los mismos datos de producto para comparar el prompt básico y el Fast Prompt.

Durante el experimento se requieren dos consultas. En un flujo productivo se utiliza solamente el prompt optimizado.

```python
if usar_api:
    resultado_basico = consultar_modelo(prompt_basico)
    resultado_fast = consultar_modelo(prompt_fast)

    print("=== RESULTADO PROMPT BÁSICO ===")
    print(resultado_basico["texto"])

    print("\n=== RESULTADO FAST PROMPT ===")
    print(resultado_fast["texto"])
else:
    print("Experimento preparado. No se realizaron consultas a la API.")
```

## 7.5 Resultados de demostración

Debido a que la ejecución mediante API requiere créditos, se conservan ejemplos de salida para mostrar la comparación. La integración real queda implementada en la notebook.

```python
resultado_basico_ejemplo = """
¡Renová tu estilo con Zapatillas Urban Street!

Descubrí unas zapatillas urbanas cómodas, con diseño moderno
y suela liviana. Ideales para acompañarte todos los días
y completar tu look urbano.

Conocé Zapatillas Urban Street y sumá estilo a tus pasos.
"""

resultado_fast_ejemplo = """
PUBLICACIÓN:
Dale un toque urbano a tu día con Zapatillas Urban Street.
Su diseño moderno, comodidad y suela liviana las convierten
en una opción ideal para quienes buscan acompañar su estilo
con un calzado pensado para el ritmo cotidiano.

CTA:
Descubrí Zapatillas Urban Street en nuestra tienda online.

HASHTAGS:
#ZapatillasUrbanas #ModaUrbana #EstiloUrbano #StreetStyle #Zapatillas
"""

print("=== RESULTADO PROMPT BÁSICO ===")
print(resultado_basico_ejemplo)

print("=== RESULTADO FAST PROMPT ===")
print(resultado_fast_ejemplo)
```

# 9. Implementación texto → imagen

La segunda parte de la solución utiliza los mismos datos del producto para crear una pieza visual.

Para evitar depender de una API de generación de imagen paga, esta POC utiliza una herramienta externa gratuita. El prompt se genera en Python y luego se copia manualmente en NightCafe u otra herramienta equivalente.

## 9.1 Fast Prompt para imagen

```python
def crear_prompt_imagen(datos):
    return f"""
ROL
Actúa como director creativo especializado en fotografía publicitaria
para e-commerce de moda urbana.

CONTEXTO
Producto: {datos['nombre']}.
Características: {datos['caracteristicas']}.
Público objetivo: {datos['publico']}.
Canal: {datos['canal']}.

TAREA
Genera una imagen publicitaria del producto para una publicación de Instagram.

ESTILO VISUAL
- Fotografía publicitaria moderna y urbana.
- El producto debe ser el elemento principal.
- Escenario urbano contemporáneo y minimalista.
- Iluminación profesional.
- Composición dinámica dirigida a un público joven.

RESTRICCIONES
- No incluir texto.
- No incluir precios ni descuentos.
- No agregar logotipos o marcas inexistentes.
- No agregar características del producto que no hayan sido proporcionadas.
- Evitar elementos que distraigan del producto.

FORMATO
Imagen cuadrada 1:1, pensada para una publicación de Instagram.
""".strip()

prompt_imagen = crear_prompt_imagen(producto)
print(prompt_imagen)
```

## 9.2 Procedimiento de generación

1. Ejecutar la celda anterior.
2. Copiar el prompt generado.
3. Abrir NightCafe u otra herramienta gratuita de generación de imágenes.
4. Pegar el prompt.
5. Generar la imagen.
6. Guardar el resultado como `zapatillas_urban_street.png`.
7. Subir la imagen al mismo repositorio que esta notebook.

# 10. Resultados

La implementación demuestra que una misma ficha de producto puede utilizarse como base para producir contenido textual y visual.

En el modelo texto → texto, la comparación muestra que el Fast Prompt presenta mayor cumplimiento que una instrucción básica. El contenido resultante respeta mejor el público objetivo, tono, CTA, hashtags y estructura solicitada.

En el modelo texto → imagen, la estructura del prompt permite definir de forma explícita el producto, público, estilo visual, restricciones y formato. Esto reduce la posibilidad de obtener una imagen poco relacionada con el objetivo comercial.

La POC logra resolver la problemática propuesta a nivel conceptual y técnico: automatiza la construcción de prompts reutilizables para dos modalidades de IA generativa y permite adaptar el flujo a diferentes productos modificando únicamente los datos de entrada.

# 11. Análisis de costos y eficiencia

La solución intenta evitar consultas innecesarias:

- Construir o visualizar prompts: **0 consultas**.

# 12. Reutilización con otro producto
El mismo sistema puede reutilizarse cambiando únicamente la ficha de datos.

```python
otro_producto = {
    "nombre": "Mochila CityPack",
    "caracteristicas": "mochila urbana compacta con compartimento acolchado para notebook",
    "publico": "estudiantes universitarios y jóvenes profesionales",
    "canal": "Instagram",
    "tono": "práctico y cercano",
    "objetivo": "presentar el producto y generar visitas a la tienda online"
}

print("PROMPT TEXTO")
print(crear_fast_prompt(otro_producto))

print("\nPROMPT IMAGEN")
print(crear_prompt_imagen(otro_producto))
```
